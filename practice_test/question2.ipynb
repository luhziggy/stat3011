```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import gaussian_kde
from KDEpy import FFTKDE
```

## Data for visualization

```python
data = pd.read_csv('https://raw.githubusercontent.com/bersavosh/S3011/main/practice_test/weather_data.csv')
budapest_humidity = data['BUDAPEST_humidity'].values
```

## Histogram

```python
fig = plt.figure(figsize=(10,7))
ax = fig.add_subplot(1,1,1)

ax.hist(
    budapest_humidity,
    bins='fd',
    density=True,
    edgecolor='black',
    alpha=0.7,
    color='skyblue',
)

# Axis labels & title
ax.set_xlabel('Humidity (%)', fontsize=12)
ax.set_ylabel('Density', fontsize=12)
ax.set_title('Histogram of Humidity in Budapest', fontsize=14)
ax.grid(True, linestyle='--', alpha=0.5)

plt.show()
```

## KDE plot

```python
fig = plt.figure(figsize=(10,7))
ax = fig.add_subplot(1,1,1)

kde = gaussian_kde(budapest_humidity)
x_grid = np.linspace(
    budapest_humidity.min() - 5, budapest_humidity.max() + 5, 1000
)
y_kde = kde(x_grid)

# Plot the density curve
ax.plot(x_grid, y_kde, color='darkblue', linewidth=2, label='KDE (SciPy)')
ax.fill_between(x_grid, y_kde, color='skyblue', alpha=0.4)

# Axis labels & title
ax.set_xlabel('Humidity (%)', fontsize=12)
ax.set_ylabel('Density', fontsize=12)
ax.set_title(
    'Kernel Density Estimation (KDE) of Humidity in Budapest', fontsize=14
)
ax.grid(True, linestyle='--', alpha=0.5)
ax.legend(fontsize=11)

plt.show()
```
