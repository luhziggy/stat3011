## Making a simple plot

Implict

````python
plt.plot(data['speed_mph_mean'])
````

Explicit

````python
fig = plt.figure(figsize=[8, 6])
ax = fig.add_subplot()
ax.plot(data['speed_mph_mean'], ':s')
ax.set_xlabel('Trip Number')
ax.set_ylabel('Mean Speed')
fig.savefig('plot.jpg', dpi=300)
````
